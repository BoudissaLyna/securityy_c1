<h3 style="color:#ff5555;margin-top:0;"> The Problem</h3>
<p style="color:#e0e0e0;">DNS tunneling is a method used by attackers to bypass security controls by encapsulating non-DNS traffic (HTTP, SSH, etc.) within DNS queries and responses. Since DNS is often <em>allowed through firewalls</em>, this creates a <strong style="color:#ffaa00;">covert channel</strong> for data exfiltration, C2 communication, and malware delivery.</p>

</ul>
</div>
<div style="background:#0d1117;padding:14px;border-radius:10px;border:1px solid #1e2a3a;">
<p style="color:#aaaaaa;font-size:13px;margin:0;"> <strong style="color:#e0e0e0;">Why DNS Tunneling?</strong> DNS is one of the most overlooked attack vectors — it's universally allowed through firewalls, rarely inspected by traditional IDS/IPS, and provides attackers a low-noise channel for exfiltrating sensitive data or maintaining persistent C2 access. Detecting it requires behavioral, statistical, and lexical analysis rather than simple signature matching.</p>

## 3. Data Preprocessing & Cleaning
<a id="3-data-preprocessing--cleaning"></a>


<div style="background:#0f172a;padding:18px;border-radius:12px;border-left:4px solid #ffaa00;margin:10px 0;">
<h3 style="color:#ffaa00;margin-top:0;"> Cleaning Pipeline Overview</h3>
<p style="color:#e0e0e0;">The <code style="background:#1e2a3a;padding:2px 6px;border-radius:4px;">process_and_merge()</code> function orchestrates all preprocessing steps into a single reproducible pipeline:</p>
<table style="width:100%;border-collapse:collapse;margin:12px 0;color:#e0e0e0;">
  <tr style="border-bottom:1px solid #ffaa0033;">
    <th style="text-align:left;padding:8px;color:#ffaa00;">Step</th>
    <th style="text-align:left;padding:8px;color:#ffaa00;">Operation</th>
    <th style="text-align:left;padding:8px;color:#ffaa00;">Why It Matters</th>
  </tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;">1</td><td style="padding:8px;"><strong>Benign Labeling</strong></td><td style="padding:8px;">Assigns <code>label=0</code> and <code>attack_type='benign'</code></td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;">2</td><td style="padding:8px;"><strong>Malicious Labeling</strong></td><td style="padding:8px;">Iterates attack folders, assigns <code>label=1</code> + tool name</td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;">3</td><td style="padding:8px;"><strong>Deduplication</strong></td><td style="padding:8px;">Removes redundant rows to prevent overfitting</td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;">4</td><td style="padding:8px;"><strong>NaN Imputation</strong></td><td style="padding:8px;">Numeric NaNs → <code>0</code>, Categorical NaNs → <code>'unknown'</code></td></tr>
  <tr><td style="padding:8px;">5</td><td style="padding:8px;"><strong>Shuffling</strong></td><td style="padding:8px;">Randomizes order to prevent temporal/sequential bias</td></tr>
</table>
</div>


```python
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
print('Pipeline environment ready.')
```

```python
DATA_DIR = "/content/data/dns-exfiltration-dataset/02_generated_dataset"

benign_file = os.path.join(DATA_DIR, "benign", "benign.csv")

malicious_base_dir = os.path.join(DATA_DIR, "malicious")

FINAL_MERGED_PATH = os.path.join(DATA_DIR, 'pipeline_final_merged.csv')


def process_and_merge():
    all_dfs = []

    # 1. Process Benign
    if os.path.exists(benign_file):
        print(f"Loading Benign: {benign_file}...")
        df = pd.read_csv(benign_file, low_memory=False)
        print(f"  Raw rows: {len(df)}")
        df.drop_duplicates(inplace=True)
        df['label'] = 0
        df['attack_type'] = 'benign'

        # Fill NaNs
        num_cols = df.select_dtypes(include=[np.number]).columns
        df[num_cols] = df[num_cols].fillna(0)
        cat_cols = df.select_dtypes(exclude=[np.number]).columns
        df[cat_cols] = df[cat_cols].fillna('unknown')

        all_dfs.append(df)
        print(f"  Rows after cleaning: {len(df)}")

    # 2. Process Malicious 
    if os.path.exists(malicious_base_dir):

        for attack_type in os.listdir(malicious_base_dir):
            attack_folder = os.path.join(malicious_base_dir, attack_type)

            f_path = os.path.join(attack_folder, f"{attack_type}.csv")

            if os.path.exists(f_path):
                print(f"Loading Malicious: {f_path}...")

                df = pd.read_csv(f_path, low_memory=False)
                print(f"  Raw rows: {len(df)}")

                df.drop_duplicates(inplace=True)
                df['label'] = 1
                df['attack_type'] = attack_type

                # Fill NaNs
                num_cols = df.select_dtypes(include=[np.number]).columns
                df[num_cols] = df[num_cols].fillna(0)
                cat_cols = df.select_dtypes(exclude=[np.number]).columns
                df[cat_cols] = df[cat_cols].fillna('unknown')

                all_dfs.append(df)
                print(f"  Rows after cleaning: {len(df)}")

    # 3. Final Merge and Shuffle (UNCHANGED)
    if all_dfs:
        print("\nMerging all cleaned datasets...")
        merged_df = pd.concat(all_dfs, ignore_index=True)
        merged_df.drop_duplicates(inplace=True)

        merged_df = merged_df.sample(frac=1, random_state=42).reset_index(drop=True)

        print(f"Final Merged Shape: {merged_df.shape}")
        merged_df.to_csv(FINAL_MERGED_PATH, index=False)
        print(f"Saved to {FINAL_MERGED_PATH}")

        return merged_df

    return None


# Execute
PIPELINE_DATA = process_and_merge()
```

```python
#working with a sample
df_sample = PIPELINE_DATA.sample(n=100000, random_state=42)

print('--- Final Merged Audit ---')
print(f'Sample Shape: {df_sample.shape}')

print('\n--- Class Balance (Sample) ---')
print(df_sample['label'].value_counts(normalize=True))

print('\n--- Attack Type Breakdown ---')
print(df_sample['attack_type'].value_counts())
# df_sample = PIPELINE_DATA
```

```python
def light_cleaning(df):
    print(f'Starting cleaning on {len(df)} rows...')
    original_len = len(df)

    # Standardize DNS strings
    if 'dns_domain_name' in df.columns:
        df['dns_domain_name'] = df['dns_domain_name'].str.lower().str.strip()

    # Handle Potential NaN residues from loading
    num_cols = df.select_dtypes(include=[np.number]).columns
    df[num_cols] = df[num_cols].fillna(0)
    cat_cols = df.select_dtypes(exclude=[np.number]).columns
    df[cat_cols] = df[cat_cols].fillna('unknown')

    print(f'Cleaning complete. Rows affected: {original_len - len(df)}')
    return df

df_sample = light_cleaning(df_sample)
```

## 5. Feature Engineering
<a id="5-feature-engineering"></a>


<div style="background:#0f172a;padding:18px;border-radius:12px;border-left:4px solid #ffaa00;margin:10px 0;">
<h3 style="color:#ffaa00;margin-top:0;"> Why Engineer Features?</h3>
<p style="color:#e0e0e0;">Raw DNS query strings contain rich information that classical ML models cannot directly use. We extract <strong>interpretable, numerical features</strong> that encode the behavioral fingerprints of tunneling activity:</p>
<div style="display:flex;gap:12px;flex-wrap:wrap;margin:12px 0;">
  <div style="flex:1;min-width:160px;background:#0d1117;padding:12px;border-radius:8px;border:1px solid #ffaa0022;">
    <strong style="color:#ffaa00;"> Shannon Entropy</strong>
    <p style="color:#aaaaaa;font-size:12px;">High entropy → random-looking → likely Base64/Hex payload. Formula: H = -Σ p·log₂(p)</p>
  </div>
  <div style="flex:1;min-width:160px;background:#0d1117;padding:12px;border-radius:8px;border:1px solid #00ffcc22;">
    <strong style="color:#00ffcc;"> Query Length</strong>
    <p style="color:#aaaaaa;font-size:12px;">Malicious queries embed data → significantly longer than benign queries</p>
  </div>
  <div style="flex:1;min-width:160px;background:#0d1117;padding:12px;border-radius:8px;border:1px solid #ff555522;">
    <strong style="color:#ff5555;"> Digit / Letter Ratios</strong>
    <p style="color:#aaaaaa;font-size:12px;">Alphanumeric encoding shifts character-class balance vs natural language</p>
  </div>
  <div style="flex:1;min-width:160px;background:#0d1117;padding:12px;border-radius:8px;border:1px solid #ffaa0022;">
    <strong style="color:#ffaa00;"> Subdomain Depth</strong>
    <p style="color:#aaaaaa;font-size:12px;">Tunneling tools often use deeply nested subdomains to encode data chunks</p>
  </div>
</div>
<p style="color:#aaaaaa;font-size:13px;">Note: Shannon entropy computation is shown but commented out below since <code>character_entropy</code> already exists in the raw dataset. The engineering step adds complementary non-redundant features.</p>
</div>


```python
def calculate_shannon_entropy(text):
    if not isinstance(text, str) or len(text) == 0: return 0
    prob = [float(text.count(c)) / len(text) for c in dict.fromkeys(list(text))]
    return - sum([p * np.log2(p) for p in prob])

def engineer_features(df):
    print('Engineering advanced features...')

    # Ensure string type
    df['dns_domain_name'] = df['dns_domain_name'].astype(str)

    # Structural Features
    df['query_length'] = df['dns_domain_name'].apply(len)
    df['num_dots'] = df['dns_domain_name'].apply(lambda x: x.count('.'))
    df['subdomain_depth'] = df['dns_domain_name'].apply(lambda x: len(x.split('.')) - 1)

    # Character-based Features
    df['digit_count'] = df['dns_domain_name'].apply(lambda x: sum(c.isdigit() for c in x))
    df['letter_count'] = df['dns_domain_name'].apply(lambda x: sum(c.isalpha() for c in x))
    df['special_char_count'] = df['dns_domain_name'].apply(lambda x: sum(not c.isalnum() and c != '.' for c in x))

    df['digit_ratio'] = df['digit_count'] / (df['query_length'] + 1)
    df['letter_ratio'] = df['letter_count'] / (df['query_length'] + 1)
    df['special_ratio'] = df['special_char_count'] / (df['query_length'] + 1)

    df['unique_chars'] = df['dns_domain_name'].apply(lambda x: len(set(x)))

    # Statistical Features
    #df['shannon_entropy'] = df['dns_domain_name'].apply(calculate_shannon_entropy)

    print(f'Feature engineering complete. New shape: {df.shape}')
    return df

df_final=engineer_features(df_sample)
```

## 6. Feature Selection & Leakage Prevention
<a id="6-feature-selection--leakage-prevention"></a>


<div style="background:#0f172a;padding:18px;border-radius:12px;border-left:4px solid #ff5555;margin:10px 0;">
<h3 style="color:#ff5555;margin-top:0;"> Critical: Data Leakage Prevention</h3>
<p style="color:#e0e0e0;">In network security datasets, certain features encode ground-truth information that a real-time model would <em>never have access to</em>. Including them produces artificially inflated metrics that fail in production. We enforce a strict drop policy:</p>
<table style="width:100%;border-collapse:collapse;margin:12px 0;color:#e0e0e0;">
  <tr style="border-bottom:1px solid #ff555533;"><th style="text-align:left;padding:8px;color:#ff5555;">Feature</th><th style="text-align:left;padding:8px;color:#ff5555;">Reason for Removal</th></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;"><code>flow_id</code></td><td style="padding:8px;">Contains embedded timestamp + host ID → direct leakage</td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;"><code>timestamp</code></td><td style="padding:8px;">Temporal ordering of captures → irrelevant for generalization</td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;"><code>src_ip / dst_ip</code></td><td style="padding:8px;">Lab-specific IPs → won't generalize to other networks</td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;"><code>dns_domain_name</code></td><td style="padding:8px;">Raw string → replaced by engineered numerical features</td></tr>
  <tr style="border-bottom:1px solid #1e2a3a;"><td style="padding:8px;"><code>uni/bi/tri_gram</code></td><td style="padding:8px;">High-cardinality n-grams → redundant after lexical engineering</td></tr>
  <tr><td style="padding:8px;"><code>attack_type</code></td><td style="padding:8px;">Sub-label → binary task uses only <code>label</code> column</td></tr>
</table>
<p style="color:#aaaaaa;font-size:13px;"> All dropped features are logged with justifications using <code>drop_with_justification()</code> for full reproducibility and audit trail.</p>
</div>


```python
justifications = {}

def drop_with_justification(df, col, reason):
    if col in df.columns:
        justifications[col] = reason
        print(f'DROPPING [{col}]: {reason}')
        return df.drop(columns=[col])
    return df
```

```python
# 1. Identifiers (Source of direct leakage)
df_final = drop_with_justification(df_final, 'flow_id', 'Direct leakage: Contains timestamp and host IDs.')
df_final = drop_with_justification(df_final, 'timestamp', 'Capture bias: Temporal ordering irrelevant for real-time detection.')
df_final = drop_with_justification(df_final, 'src_ip', 'Leakage: Host-specific IP addresses from lab capture.')
df_final = drop_with_justification(df_final, 'dst_ip', 'Leakage: Destination IPs tied to specific lab DNS servers.')

# 2. Raw Strings (Already engineered)
df_final = drop_with_justification(df_final, 'dns_domain_name', 'Redundant: Structural/Lexical features derived.')
df_final = drop_with_justification(df_final, 'uni_gram_domain_name', 'Redundant/High Cardinality.')
df_final = drop_with_justification(df_final, 'bi_gram_domain_name', 'Redundant/High Cardinality.')
df_final = drop_with_justification(df_final, 'tri_gram_domain_name', 'Redundant/High Cardinality.')
df_final = drop_with_justification(df_final, 'character_distribution', 'Complex format: Shannon entropy used instead.')

# 3. Sub-labels
df_final = drop_with_justification(df_final, 'attack_type', 'Detailed label: Binary classification uses [label] column.')
df_final = drop_with_justification(df_final, 'dns_top_level_domain', 'Categorical bias: Lab domains only.')
df_final = drop_with_justification(df_final, 'dns_second_level_domain', 'Categorical bias: Lab domains only.')
```

### 6.1 Categorical Encoding


<div style="background:#0f172a;padding:14px;border-radius:10px;border-left:3px solid #00ffcc;margin:8px 0;">
<p style="color:#e0e0e0;font-size:13px;margin:0;">Categorical features (DNS record types, protocol classes) must be converted to integers for ML models. We use <strong style="color:#ffaa00;">Label Encoding</strong> which maps each category to an integer. This is preferred over One-Hot Encoding for tree-based models (RF, XGBoost, GBM) because: it avoids high-dimensional sparse matrices, and decision trees can capture ordinal relationships natively. For linear models, OHE would be more appropriate, but since our ensemble models dominate, LE is the right choice.</p>
</div>


```python
def encode_categoricals(df):
    print('Encoding categorical columns...')
    le = LabelEncoder()

    cat_cols = df.select_dtypes(exclude=[np.number]).columns
    for col in cat_cols:
        print(f'  Encoding [{col}] (LabelEncoding preferred for tree models)')
        df[col] = le.fit_transform(df[col].astype(str))

    return df
df_final=encode_categoricals(df_final)

```

## 8. Cross-Validation & Learning Curves
<a id="8-cross-validation"></a>


<div style="background:#0f172a;padding:18px;border-radius:12px;border-left:4px solid #ffaa00;margin:10px 0;">
<h3 style="color:#ffaa00;margin-top:0;"> 5-Fold Stratified Cross-Validation</h3>
<p style="color:#e0e0e0;">We use <strong style="color:#00ffcc;">StratifiedKFold (k=5)</strong> which ensures each fold maintains the original class ratio — critical when dealing with imbalanced security datasets. We monitor 6 metrics simultaneously:</p>
<div style="display:flex;gap:10px;flex-wrap:wrap;margin:12px 0;">
  <span style="background:#0d1117;color:#00ffcc;padding:6px 12px;border-radius:6px;font-size:13px;">Accuracy</span>
  <span style="background:#0d1117;color:#00ffcc;padding:6px 12px;border-radius:6px;font-size:13px;">F1 Score</span>
  <span style="background:#0d1117;color:#ffaa00;padding:6px 12px;border-radius:6px;font-size:13px;">ROC-AUC</span>
  <span style="background:#0d1117;color:#ffaa00;padding:6px 12px;border-radius:6px;font-size:13px;">Average Precision</span>
  <span style="background:#0d1117;color:#ff5555;padding:6px 12px;border-radius:6px;font-size:13px;">Precision</span>
  <span style="background:#0d1117;color:#ff5555;padding:6px 12px;border-radius:6px;font-size:13px;">Recall</span>
</div>
<p style="color:#e0e0e0;">For each metric, we also compute <code>train_score - val_score</code> (overfit gap). Gaps &gt; 0.05 flag severe overfitting; &gt; 0.03 flag mild overfitting. Both train and validation scores are returned via <code>return_train_score=True</code>.</p>
</div>


```python

CV = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

SCORING = ("accuracy", "f1", "roc_auc",
           "precision", "recall", "average_precision")

cv_results = {}
for name, pipe in pipelines.items():
    t0 = time.time()
    res = cross_validate(pipe, X_train, y_train,
                         cv=CV, scoring=SCORING,
                         return_train_score=True, n_jobs=-1)
    elapsed = time.time() - t0

    cv_results[name] = res
    print(f"\n  [{name}]  ({elapsed:.1f}s)")
    for metric in ("accuracy", "f1", "roc_auc"):
        val_m  = res[f"test_{metric}"].mean()
        val_s  = res[f"test_{metric}"].std()
        trn_m  = res[f"train_{metric}"].mean()
        gap    = trn_m - val_m
        
        if gap > 0.05:
            flag = "  ❌ OVERFIT (severe)"
        elif gap > 0.03:
            flag = "  ⚠  OVERFIT (mild)"
        else:
            flag = "  ✓"
        print(f"    {metric:15s}  train={trn_m:.4f}  val={val_m:.4f}±{val_s:.4f}"
              f"  gap={gap:.4f}{flag}")
```

### 8.1 Learning Curves
<div style="background:#0f172a;padding:14px;border-radius:10px;border-left:3px solid #00ffcc;margin:8px 0;">
<p style="color:#e0e0e0;font-size:13px;margin:0;">Learning curves plot ROC-AUC as a function of training set size, diagnosing the <strong style="color:#ffaa00;">bias-variance trade-off</strong> for each model:
<br>• <strong style="color:#00ffcc;">Converging curves</strong> → Good fit — model generalizes well
<br>• <strong style="color:#ff5555;">Large gap (train >> val)</strong> → Overfitting — needs more regularization or data
<br>• <strong style="color:#ffaa00;">Both curves low & converged</strong> → Underfitting — model needs more capacity
<br>The shaded bands show ±1 standard deviation across folds.</p>
</div>


```python
fig, axes = plt.subplots(1, len(pipelines), figsize=(5 * len(pipelines), 4))
TRAIN_SIZES = np.linspace(0.10, 1.0, 8)

for ax, (name, pipe) in zip(axes, pipelines.items()):
    train_sz, train_sc, val_sc = learning_curve(
        pipe, X_train, y_train,
        cv=CV, scoring="roc_auc",
        train_sizes=TRAIN_SIZES,
        n_jobs=-1)

    ax.plot(train_sz, train_sc.mean(axis=1), "o-",  label="Train AUC")
    ax.fill_between(train_sz,
                    train_sc.mean(1) - train_sc.std(1),
                    train_sc.mean(1) + train_sc.std(1), alpha=0.15)
    ax.plot(train_sz, val_sc.mean(axis=1),   "s--", label="Val AUC")
    ax.fill_between(train_sz,
                    val_sc.mean(1) - val_sc.std(1),
                    val_sc.mean(1) + val_sc.std(1), alpha=0.15)
    ax.set_title(name); ax.set_xlabel("Training samples")
    ax.set_ylabel("ROC-AUC"); ax.legend(); ax.grid(True, alpha=0.3)
    ax.set_ylim(0.5, 1.02)

os.makedirs("outputs", exist_ok=True) 
plt.suptitle("Learning Curves — DNS Tunneling Detection", fontsize=13)
plt.tight_layout()
plt.savefig("outputs/learning_curves.png", dpi=150)
plt.show()
```

It is then refit on the <strong style="color:#00ffcc;">entire training set</strong> (all 5 folds combined) before being evaluated on the held-out test set. ### 9.2 Permutation Feature Importance
<div style="background:#0f172a;padding:14px;border-radius:10px;border-left:3px solid #ffaa00;margin:8px 0;">
<p style="color:#e0e0e0;font-size:13px;margin:0;">Permutation importance measures how much the ROC-AUC drops when a feature's values are randomly shuffled (breaking the feature-label relationship). This is more reliable than tree-based impurity importance because it:
<br>• Works on any model (not just tree-based)
<br>• Accounts for feature interactions (correlated features share importance)
<br>• Uses the <strong>test set</strong> → reflects real-world impact, not training memorization
<br>Error bars (±std) show stability across 15 repetitions.</p>
</div>


```python
perm = permutation_importance(best_pipe, X_test, y_test,
                               n_repeats=15, scoring="roc_auc",
                               random_state=SEED, n_jobs=-1)

imp_df = pd.DataFrame({
    "feature"    : FEATURES,
    "importance" : perm.importances_mean,
    "std"        : perm.importances_std,
}).sort_values("importance", ascending=False)

print(imp_df.head(15).to_string(index=False))

fig, ax = plt.subplots(figsize=(10, 8))
top = imp_df.head(20)
ax.barh(top["feature"][::-1], top["importance"][::-1],
        xerr=top["std"][::-1], capsize=3, color="steelblue")
ax.set_xlabel("Mean AUC decrease"); ax.set_title("Top-20 Permutation Feature Importances")
ax.grid(True, axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig("outputs/feature_importance.png", dpi=150)
plt.show()
```
